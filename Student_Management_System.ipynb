```python
#GLOBAL LIST TO STORE THE DICTONARY
students = []
```

```python
#FUNCTION FOR ADD NEW STUDENTS
def add_student():
    try:
        student_id = int(input("Enter roll number: "))
    except ValueError:
        print("Invalid roll number. Please enter an integer.")
        return

    name = input("Enter name: ").strip()

    try:
        age = int(input("Enter age: "))
    except ValueError:
        print("Invalid age. Please enter an integer.")
        return

    course = input("Enter course: ").strip()
    marks_input = input("Enter numbers separated by spaces: ")

    marks = []
    for item in marks_input.split():
        try:
            marks.append(int(item))
        except ValueError:
            print(f"Skipping '{item}' (not a valid mark)")

    student = {
        'id': student_id,
        'name': name,
        'age': age,
        'course': course,
        'marks': marks
    }
    students.append(student)
    print("Student added successfully!")
```

```python
#FUNCTION FOR DISPLAY ALL STUDENTS IN LIST
def view_students():
    for student in students:
        print(f"ID:     {student['id']}")
        print(f"Name:   {student['name']}")
        print(f"Age:    {student['age']}")
        print(f"Course: {student['course']}")
        print(f"Marks:  {student['marks']}")
        print("*****************************")
```

```python
#FUNCTION FOR SEARCH PERTICULAR STUDENT
def records(search_term):
    found = False
    search_term = search_term.strip()

    for student in students:
        # Check matching ID (if search term is a number) or matching Name (case-insensitive)
        if (search_term.isdigit() and student['id'] == int(search_term)) or \
           (student['name'].lower() == search_term.lower()):
            print(f"ID:     {student['id']}")
            print(f"Name:   {student['name']}")
            print(f"Age:    {student['age']}")
            print(f"Course: {student['course']}")
            print(f"Marks:  {student['marks']}")
            print("*****************************")
            found = True

    if not found:
        print(f"No student record found matching '{search_term}'.")
```

```python
#FUNTION FOR DELETE STUDENTS RECORD
def delete_student(student_id):
    global students
    initial_length = len(students)
    students = [student for student in students if student['id'] != student_id]

    if len(students) < initial_length:
        print(f"Student with ID {student_id} deleted successfully.")
    else:
        print(f"Student with ID {student_id} not found.")
```

```python
#MENU OPTION
while True:
    print("\n---- Student Management Menu ----")
    print("1. Add student")
    print("2. View all students")
    print("3. Search by name or id")
    print("4. Delete by roll number")
    print("5. Exit")
    print("--------------------------------")

    choice = input("Enter your choice (1-5): ").strip()

    if choice == '1':
        add_student()
    elif choice == '2':
        if not students:
            print("The list is empty!")
        else:
            view_students()
    elif choice == '3':
        if not students:
            print("The list is empty!")
        else:
            search_query = input("Enter the student name or ID: ")
            records(search_query)
    elif choice == '4':
        if not students:
            print("The list is empty! No students to delete.")
        else:
            try:
                student_id_to_delete = int(input("Enter the ID of the student to delete: "))
                delete_student(student_id_to_delete)
            except ValueError:
                print("Invalid input! Please enter a numeric student ID.")
    elif choice == '5':
        print("Exiting Student Management System.")
        break
    else:
        print("Invalid choice. Please enter a number between 1 and 5.")
```
